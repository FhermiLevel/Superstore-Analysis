```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"])
```

```python
df.info()
```

```python
df[["Country", "City"]] = df["Country, City"].str.split(",", expand=True)
```

```python
df["Country"] = df["Country"].str.strip()
df["City"] = df["City"].str.strip()
```

```python
df.head()
```

```python
column_index = df.columns.get_loc("Country, City")
country = df.pop("Country")
city = df.pop("City")
```

```python
column_index 
```

```python
df.insert(column_index + 1, "Country", country)
df.insert(column_index + 2, "City", city)
```

```python
df.drop(columns="Country, City", inplace=True)
```

```python
df.head(2)
```

```python
df["Order _ID"].nunique()
```

```python
df.rename(columns={"Order _ID":"Order_ID"}, inplace=True)
```

```python
df["Profit_Margin(%)"] = (df["Profit"] / df["Sales"]) * 100   #Creating a profit margin column
```

```python
df.head(2)
```

```python
df["Days_to_ship"] = (df["Ship_Date"] - df["Order_Date"]).dt.days
```

```python
df.head()
```

```python
df[df["Ship_Date"] < df["Order_Date"]]
```

```python
df["Invalid_shipment"] = df["Ship_Date"] < df["Order_Date"]
```

```python
df.head(2)
```

```python
invalid_shipment_pct =(df["Ship_Date"] < df["Order_Date"]).mean() * 100

invalid_shipment_pct
```

```python
df = df[df["Ship_Date"] >= df["Order_Date"]]
```

```python
df.info()
```

```python
df.head()
```

```python
df["Weekdays"] = df["Order_Date"].dt.day_name()
df["month_name"] = df["Order_Date"].dt.month_name()
df["Quarters"] = "Q" + df["Order_Date"].dt.quarter.astype(str)
```

### Just trying out some visuals here, you can get all visuals on Power BI file attached to this project.

```python
state_count = df["State"].value_counts().head(10)
state_count 
```

```python
plt.bar(state_count.index, state_count, color= "b")
plt.xticks(rotation = 90)
plt.xlabel("States")
plt.ylabel("Count of Sales")

plt.show()
```

```python
sub_cat_profit = df.groupby("Sub_Category")["Profit"].sum().reset_index(name= "Total Profit")
sub_cat_profit.sort_values(by = "Total Profit", ascending=False, inplace= True)
```

```python
plt.bar(sub_cat_profit["Sub_Category"], sub_cat_profit["Total Profit"])
plt.xticks(rotation = 90)
plt.xlabel("Sub_Categories")
plt.ylabel("Total Profit")
plt.title("Sub Category by Profit")
plt.show()
```
